1) Update the weights of given neural network using perceptron learning for one iteration

```python
import numpy as np

# inputs
x = np.array([[0.1],
            [0.2],
            [0.3],
            [0.4]])


#eta 
n=0.1

# weights
w = np.array([[0.25],
            [-0.2],
            [-0.3],
            [0.4]]
            )

# net
net = w.T @ x

# output
output = np.sign(net)

# desired output
d=1

# error
dw = n*(d-output)*x

# updated weight
w=w+dw

# print values
print("Updated Weights")
print(np.round(np.ravel(w), 5))
```

2) Back Propagation of ax^2 - b =0; given weights, a=1, b=7; eta=0.5

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))

def d_sigmoid(x):
    return x*(1-x)
```

```python
# inputs and target
a, b = 2.0, 7.0
x=np.array([[a] ,
            [b]])

# desired values
root_val = np.sqrt(b/a)
t = np.array([[root_val],
     [-root_val]])

# Initial Weights
W_ji = np.array([[-0.1, 0.9, -0.2],
                [0.8, -0.3, 0.1]])

W_kj = np.array([[-0.2, 0.4],
                [0.8, -0.6 ],
                [-0.5, 0.9]])

eta =0.5
iterations=2
```

BPA

```python
for it in range(1, iterations+1):

    # Forward Pass

    net_j = np.dot(W_ji.T, x)
    O_j = sigmoid(net_j)

    net_k = np.dot(W_kj.T, O_j) 
    O_k = sigmoid(net_k)


    # BackWard Pass

    delta_k = (t- O_k)*d_sigmoid(O_k)
    dW_kj = eta * O_j @ delta_k.T

    delta_j = np.dot(W_kj, delta_k) * d_sigmoid(O_j)
    dW_ji = eta * np.dot(x, delta_j.T)


    # Updated Weights
    W_kj +=dW_kj
    W_ji +=dW_ji

    # Print Values
    print(f"--- Iteration {it} ---")
    print(f"Outputs (O_k):\n{O_k.ravel()}")
    print(f"Updated W_kj:\n{W_kj}")
    print(f"Updated W_ji:\n{W_ji}\n")
```
