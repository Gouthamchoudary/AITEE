## Perceptron Learning

Given an input vector $\mathbf{x} = \begin{bmatrix} 0.2 \\ 0.4 \\ 0.5 \end{bmatrix}$ and weight matrix $W = \begin{bmatrix} 0.2 & 0.1 & -0.2 \\ -0.4 & -0.35 & 0.2 \\ -0.4 & 0.4 & -0.4 \end{bmatrix}$:

Compute the net input vector $\mathbf{z} = W^T \mathbf{x}$ and the activation output vector $\mathbf{y} = \sigma(\mathbf{z})$ using the sigmoid function $\sigma(z) = \frac{1}{1 + e^{-z}}$.

Write a NumPy script to calculate and print both $\mathbf{z}$ and $\mathbf{y}$ rounded to 5 decimal places.

<details>
<summary><b>Solution</b></summary>

<br>

**Step 1: Net Input Calculation**

The net input vector $\mathbf{z}$ is calculated using matrix multiplication:
$$\mathbf{z} = W^T \mathbf{x}$$

Substituting $W^T$ and $\mathbf{x}$:
$$\mathbf{z} = \begin{bmatrix} 0.2 & -0.4 & -0.4 \\ 0.1 & -0.35 & 0.4 \\ -0.2 & 0.2 & -0.4 \end{bmatrix} \begin{bmatrix} 0.2 \\ 0.4 \\ 0.5 \end{bmatrix}$$

Performing row-by-column multiplication:
* $z_1 = (0.2)(0.2) + (-0.4)(0.4) + (-0.4)(0.5) = 0.04 - 0.16 - 0.20 = \mathbf{-0.32}$
* $z_2 = (0.1)(0.2) + (-0.35)(0.4) + (0.4)(0.5) = 0.02 - 0.14 + 0.20 = \mathbf{0.08}$
* $z_3 = (-0.2)(0.2) + (0.2)(0.4) + (-0.4)(0.5) = -0.04 + 0.08 - 0.20 = \mathbf{-0.16}$

$$\mathbf{z} = \begin{bmatrix} -0.32 \\ 0.08 \\ -0.16 \end{bmatrix}$$

---

**Step 2: Sigmoid Activation Function**

Apply the sigmoid function $\sigma(z) = \frac{1}{1 + e^{-z}}$ element-wise to $\mathbf{z}$:

* $y_1 = \sigma(-0.32) = \frac{1}{1 + e^{0.32}} \approx \mathbf{0.42067}$
* $y_2 = \sigma(0.08) = \frac{1}{1 + e^{-0.08}} \approx \mathbf{0.51999}$
* $y_3 = \sigma(-0.16) = \frac{1}{1 + e^{0.16}} \approx \mathbf{0.46008}$

---

**Final Output Vector**

$$\mathbf{y} = \begin{bmatrix} 0.42067 \\ 0.51999 \\ 0.46008 \end{bmatrix}$$

</details>

```python

import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def main():

    # input vector
    x = np.array([[0.2],
                  [0.4],
                  [0.5]])


    # weights
    w = np.array([[0.2, 0.1, -0.2],
              [-0.4, -0.35, 0.2],
              [-0.4, 0.4, -0.4]])

    #calculate Net
    net = w.T @x

    #output
    output =sigmoid(net)


    # Clean formatted printing using vectorized operations
    print("\n Net Inputs (z): ---")
    print(np.round(net.ravel(), 5))

    print("\n \nActivation Outputs (y): ---")
    print(np.round(output.ravel(), 5))


if __name__ == "__main__":
    main()
```
