## Perceptron Learning

Given an input vector $\mathbf{x} = \begin{bmatrix} 0.2 \\ 0.4 \\ 0.5 \end{bmatrix}$ and weight matrix $W = \begin{bmatrix} 0.2 & 0.1 & -0.2 \\ -0.4 & -0.35 & 0.2 \\ -0.4 & 0.4 & -0.4 \end{bmatrix}$:

Compute the net input vector $\mathbf{z} = W^T \mathbf{x}$ and the activation output vector $\mathbf{y} = \sigma(\mathbf{z})$ using the sigmoid function $\sigma(z) = \frac{1}{1 + e^{-z}}$.

Write a NumPy script to calculate and print both $\mathbf{z}$ and $\mathbf{y}$ rounded to 5 decimal places.

<details>
<summary><b> Click for Solution</b></summary>

<br>

**Step 1: Net Input Calculation**

The net input vector $\mathbf{z}$ is calculated using matrix multiplication:
$$\mathbf{z} = W^T \mathbf{x}$$

Substituting $W^T$ and $\mathbf{x}$:
$$\mathbf{z} = \begin{bmatrix} 0.2 & -0.4 & -0.4 \\ 0.1 & -0.35 & 0.4 \\ -0.2 & 0.2 & -0.4 \end{bmatrix} \begin{bmatrix} 0.2 \\ 0.4 \\ 0.5 \end{bmatrix}$$

Performing row-by-column multiplication:
* $z_1 = (0.2)(0.2) + (-0.4)(0.4) + (-0.4)(0.5) = 0.04 - 0.16 - 0.20 = \mathbf{-0.32}$
* $z_2 = (0.1)(0.2) + (-0.35)(0.4) + (0.4)(0.5) = 0.02 - 0.14 + 0.20 = \mathbf{0.08}$
* $z_3 = (-0.2)(0.2) + (0.2)(0.4) + (-0.4)(0.5) = -0.04 + 0.08 - 0.20 = \mathbf{-0.16}$

$$\mathbf{z} = \begin{bmatrix} -0.32 \\ 0.08 \\ -0.16 \end{bmatrix}$$

---

**Step 2: Sigmoid Activation Function**

Apply the sigmoid function $\sigma(z) = \frac{1}{1 + e^{-z}}$ element-wise to $\mathbf{z}$:

* $y_1 = \sigma(-0.32) = \frac{1}{1 + e^{0.32}} \approx \mathbf{0.42067}$
* $y_2 = \sigma(0.08) = \frac{1}{1 + e^{-0.08}} \approx \mathbf{0.51999}$
* $y_3 = \sigma(-0.16) = \frac{1}{1 + e^{0.16}} \approx \mathbf{0.46008}$

---

**Final Output Vector**

$$\mathbf{y} = \begin{bmatrix} 0.42067 \\ 0.51999 \\ 0.46008 \end{bmatrix}$$

</details>

```python

import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def main():

    # input vector
    x = np.array([[0.2],
                  [0.4],
                  [0.5]])


    # weights
    w = np.array([[0.2, 0.1, -0.2],
              [-0.4, -0.35, 0.2],
              [-0.4, 0.4, -0.4]])

    #calculate Net
    net = w.T @x

    #output
    output =sigmoid(net)


    # Clean formatted printing using vectorized operations
    print("\n Net Inputs (z): ---")
    print(np.round(net.ravel(), 5))

    print("\n \nActivation Outputs (y): ---")
    print(np.round(output.ravel(), 5))


if __name__ == "__main__":
    main()
```

## Delta Learning


Given a single-layer perceptron with a learning rate $\eta = 0.5$, initial weight vector $w = \begin{bmatrix} -0.3 \\ 0.2 \\ -0.1 \end{bmatrix}$, and sigmoid activation function $\sigma(z) = \frac{1}{1 + e^{-z}}$:

Train the network using the Delta Rule for two input patterns:
* **Pattern 1:** $x_1 = \begin{bmatrix} -0.25 \\ 0.10 \\ -0.20 \end{bmatrix}$, target $d_1 = 0.95$
* **Pattern 2:** $x_2 = \begin{bmatrix} 0.15 \\ -0.30 \\ 0.10 \end{bmatrix}$, target $d_2 = 0.90$

Update the weight vector sequentially after each pattern.

<details>
<summary><b>Click Mathematical Solution</b></summary>

<br>

**1. Activation & Delta Rule Formulas**
* Net input: $z = w^T x$
* Output: $y = \sigma(z) = \frac{1}{1 + e^{-z}}$
* Derivative: $f'(z) = y(1 - y)$
* Error gradient: $\delta = (d - y) \cdot f'(z)$
* Weight Update: $w_{\text{new}} = w_{\text{old}} + \eta \cdot \delta \cdot x$

---

**2. Pattern 1 Update**
* $z_1 = (-0.3)(-0.25) + (0.2)(0.10) + (-0.1)(-0.20) = 0.075 + 0.02 + 0.02 = \mathbf{0.115}$
* $y_1 = \sigma(0.115) = \frac{1}{1 + e^{-0.115}} \approx \mathbf{0.52872}$
* $\delta_1 = (0.95 - 0.52872) \cdot (0.52872)(1 - 0.52872) \approx \mathbf{0.10497}$
* $w^{(1)} = \begin{bmatrix} -0.3 \\ 0.2 \\ -0.1 \end{bmatrix} + 0.5 \cdot 0.10497 \cdot \begin{bmatrix} -0.25 \\ 0.10 \\ -0.20 \end{bmatrix} = \mathbf{\begin{bmatrix} -0.31312 \\ 0.20525 \\ -0.11050 \end{bmatrix}}$

---

**3. Pattern 2 Update**
* $z_2 = (-0.31312)(0.15) + (0.20525)(-0.30) + (-0.11050)(0.10) \approx \mathbf{-0.11959}$
* $y_2 = \sigma(-0.11959) \approx \mathbf{0.47014}$
* $\delta_2 = (0.90 - 0.47014) \cdot (0.47014)(1 - 0.47014) \approx \mathbf{0.10709}$
* $w^{(2)} = w^{(1)} + 0.5 \cdot 0.10709 \cdot \begin{bmatrix} 0.15 \\ -0.30 \\ 0.10 \end{bmatrix} = \mathbf{\begin{bmatrix} -0.30509 \\ 0.18918 \\ -0.10514 \end{bmatrix}}$

</details>

Helper Functions

```python
import numpy as np

def sigmoid(z):
    return 1/(1+np.exp(-z))

def sigmoid_derivative(y):
    return y*(1-y)
```

epoch = 1

```python
def run_1_epoch(w , patterns, eta):
    print("epoch =1")

    for i, (x,d) in enumerate(patterns, start=1):
        z=np.dot(w,x)

        y=sigmoid(z)

        error = d-y
        delta = error*sigmoid_derivative(y)

        dw=eta*(delta)*x
        w=w+dw

        print(f"Pattern{i}")
        print (f" z = {z:.5f}")
        print (f" y = {y:.5f}")
        print (f" weights= {np.round(w,5)} \n")
```

epoch = 100

```python
def run_100_epochs(w,patterns, eta, num_epochs=100):
    print(f"{num_epochs} epochs")

    for epoch in range(1, num_epochs + 1):
        total_error = 0.0

        for x, d in patterns:
            z = np.dot(w, x)
            y = sigmoid(z)
            error = d - y
            delta = error * sigmoid_derivative(y)
            w = w + eta * delta * x
            total_error += abs(error)
            
        if epoch == 1 or epoch % 20 == 0 or epoch == num_epochs:
            print(
                f"Epoch {epoch:3d}/{num_epochs} | Total Error: {total_error:.5f} | Weights: {np.round(w, 5)}"
            )
    print("\n")
            
```

train until tolerance = 0.001

```python
def run_until_tolerance(w, patterns, eta, tol=0.001, max_epochs=10000):
    print("tolerance epochs")
    epoch =0

    while epoch < max_epochs:
        epoch += 1
        max_error = 0.0

        for x, d in patterns:
            z = np.dot(w, x)
            y = sigmoid(z)
            error = d - y
            delta = error * sigmoid_derivative(y)
            w = w + eta * delta * x

            if abs(error) > max_error:
                max_error = abs(error)

        if max_error < tol:
            break

    print(f"Converged in {epoch} epochs!")
    print(f"Final Max Error: {max_error:.5f}")
    print(f"Final Weights  : {np.round(w, 5)}\n")
    
```

Main Function

```python
eta = 0.5
w_initial = np.array([-0.3, 0.2, -0.1])

patterns = [
    (np.array([-0.25, 0.10, -0.20]), 0.95),
    (np.array([0.15, -0.30, 0.10]), 0.90),
]

run_1_epoch(w_initial.copy(), patterns, eta)

run_100_epochs(w_initial.copy(), patterns, eta, num_epochs=100)

run_until_tolerance(w_initial.copy(), patterns, eta, tol=0.001)
```
